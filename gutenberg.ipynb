```python
import requests as req
from bs4 import BeautifulSoup as bs
import re
import os
import time

url = "https://www.gutenberg.org/browse/languages/zh"

res = req.get(url)

soup = bs(res.text, "lxml")

# 建立 list 來放置列表資訊
list_books = []


#正則表達式
chinese = re.compile(r'^[\u4E00-\u9FFF]+$')

# 清空放置列表資訊的變數
list_books.clear()

for o in soup.select("li.pgdbetext > a[href]"):
    title = o.get_text().strip()
    if chinese.match(title):
        list_books.append({
            'name':title,
            'link':o['href']
        })

#建立資料夾
os.makedirs("project_gutenberg", exist_ok=True)

#下載文字檔
for book in list_books:
    match = re.search(r'\/ebooks\/(\d+)', book['link'])
    if not match:
        continue
    book_id = match.group(1)
    title = book['name']

    # 清理書名作為檔案名稱
    clear_title = re.sub(r'[\\/*?:"<>|\r\n]', "", title).strip()
    filename = f'project_gutenberg/{clear_title}.txt'

    # 清理後的書名為空，跳過這本書
    if not clear_title:
        continue

    # 檔案重複就跳過
    if os.path.exists(filename):
        continue    

    txt_url = [f'https://www.gutenberg.org/cache/epub/{book_id}/pg{book_id}.txt']

    

    for t_url in txt_url:
        rs = req.get(t_url)
        if rs.status_code == 200:
            text = rs.text

            #篩選中文
            chinese_text = re.sub(r'[^\u4e00-\u9fff]+', '', text)

            with open(filename, 'w', encoding='utf-8') as f :
                f.write(chinese_text)
                print(f'已完成')
                time.sleep(2)
                break
    else:
        print(f'無法下載')
```
